```python
%sql
CREATE OR REPLACE VIEW gold_person_demographics AS
SELECT
    *,
    FLOOR(months_between(current_date(), DOB) / 12) AS age
FROM
    workspace.default.silver_persons
```

```python
%sql
create or replace view gold_household_count as
select
    count(*) as household_count
from
    workspace.default.silver_households
```

```python
%sql
create or replace view gold_age_breakdown as
select 
    case 
      when age < 18 then 'under 18'
      when age between 18 and 24 then '18-24'
      when age between 25 and 34 then '25-34'
      when age between 35 and 44 then '35-44'
      when age between 45 and 54 then '45-54'
      when age between 55 and 64 then '55-64'
      when age > 64 then 'senior'
    end as age_group,
    count(*) as person_count
from
    workspace.default.gold_person_demographics
group by 1;
```

```python
%sql
CREATE OR REPLACE VIEW gold_minor_count AS
SELECT
    COUNT(*) AS minor_count
FROM gold_person_demographics
WHERE age < 18;
```

```python
%sql
CREATE OR REPLACE VIEW gold_senior_count AS
SELECT
    COUNT(*) AS senior_count
FROM gold_person_demographics
WHERE age >= 65;
```
