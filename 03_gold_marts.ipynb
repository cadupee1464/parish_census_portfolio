```python
%sql
CREATE OR REPLACE VIEW gold_person_demographics AS
SELECT
    *,
    FLOOR(months_between(current_date(), DOB) / 12) AS age
FROM
    workspace.default.silver_persons
```

```python
%sql
create or replace view gold_age_breakdown as
select 
    case 
      when age <= 5 then '0-5'
      when age between 6 and 12 then '6-12'
      when age between 13 and 19 then '13-19'
      when age between 20 and 35 then '20-34'
      when age between 36 and 50 then '35-50'
      when age between 51 and 65 then '51-65'
      when age >= 66 then '66+'
    end as age_group,
    count(*) as person_count
from
    workspace.default.gold_person_demographics
group by 1;
```

```python
%sql
CREATE OR REPLACE VIEW contributing_membership AS

SELECT COUNT (DISTINCT household_key) as contributing_units,
FROM workspace.default.gold_person_demographics
WHERE Tithing = true
    OR Volunteer = true;
```
